```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
lst = [1, 2 ,3 ,4 ,5 ,6 ,7]
array = np.array(lst)
array

```

```python
engineering_branches = ['mechanichal' , 'computerscience' , 'IT' , 'civil' , 'electrical' , 'aerospace' , 'chemical' , 'biotechnology']
data_frame = {"mechanichal" : ["CAD" , "Drawings" , "Thermodynamics" , "Mechanics"] ,
"computerscience" : ["Python" , "Java" , "Data Structures" , "Algorithms"] , 
"IT" : ["Networking" , "Hardware" , "Software Engineering" , "Database Management"] ,
"civil" : ["Structural Analysis" , "Construction Management" , "Geotechnical Engineering" , "Transportation Engineering"] ,
"electrical" : ["Circuit Design" , "Power Systems" , "Electromagnetics" , "Control Systems"] ,
"aerospace" : ["Aerodynamics" , "Propulsion" , "Flight Dynamics" , "Aerothermodynamics"] ,
"chemical" : ["Process Engineering" , "Reactions" , "Mass Transfer" , "Heat Transfer"] ,
 "biotechnology" : ["Genetics" , "Biochemistry" , "Molecular Biology" , "Bioprocess Engineering"]}
df = pd.DataFrame(data_frame)
df
```

```python
interest_domains = ["Automobile" , "Software" , "IT" , "Construction" , "Power" , "Aerospace" , "Chemical" , "Biotechnology"]
industries = ["AI" , "ML" , "Data Science" , "Civil Engineering" , "Electrical Engineering" , "Aerospace Engineering" , "Chemical Engineering" , "Biotechnology"]
#generating intern recommendations based on the interest domains and industries
recommendations = {
     "Automobile" : "Automobie intern" , 
     "software"   : "Software Developer intern" , 
     "IT" : "IT Support intern" , 
     "Construction" : "Construction Management intern" ,
     "Power" : "Power Systems intern" , 
     "Aerospace" : "Aerospace Engineering intern" , 
     "Chemical" : "Chemical Engineering intern" , 
     "Biotechnology" : "Biotechnology intern"


}
```

```python
data = []
for i in range(1,10001):
    branch = random.choice(engineering_branches)
    skills = random.sample(data_frame.get(branch , []), min(2 , len([data_frame.get])))
    interst = random.choice(interest_domains)
    cgpa = round(random.uniform(6.0 , 10.0) , 2)
    industry = random.choice(industries)
    recommendation = recommendations.get(interst , "No recommendation")

    data.append({
     "branch" : branch ,
     "skills" : skills ,
     "interest" : interst ,
     "cgpa" : cgpa ,
     "preferred_industry" : industry ,
     "industry_recommendation" : recommendation
})
df = pd.DataFrame(data)
df
```
