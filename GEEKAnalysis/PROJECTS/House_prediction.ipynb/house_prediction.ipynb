```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import fetch_california_housing
data = fetch_california_housing()
data
```

```python
dataset = pd.DataFrame(data.data , columns= data.feature_names)
type(dataset)
```

```python
dataset["price"] = data.target
dataset.head()
```

```python
corr = dataset.corr(numeric_only = True)

plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.boxplot(dataset)
#sns.pairplot(dataset)
```

```python
X = dataset.iloc[ : , : -1]
y = dataset.iloc[ : , -1]
dataset.head()
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split( X , y , test_size=0.3 , random_state= 42)
x_test
```

```python
x_test.shape , x_train.shape , y_test.shape , y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_norm = scaler.fit_transform(x_train)
x_test_norm = scaler.transform(x_test)
sns.boxplot(x_train_norm)
```

```python
x_train_norm.shape
```

```python
y_train.shape
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(x_train_norm , y_train )
```

```python
print(regression.coef_) , print(regression.intercept_)
```

```python
reg_pred = regression.predict(x_test_norm)
reg_pred
```

```python
residual = y_test - reg_pred
residual
```

```python
sns.displot(data = residual , kind= "kde")
```
